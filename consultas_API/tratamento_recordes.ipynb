```python
import pandas as pd 
```

```python
df1 = pd.read_csv("output/recordes_male.csv")
```

```python
df = df1.copy()
```

```python
df = df.loc[df["Event"] == "100 FR LCM"]
df.head()
```

```python
df["Year"] = df["Swim Date"].str[-4:]
df["Year"] = pd.to_numeric(df["Year"])
df = df.loc[df["Year"] >1970]
df.drop(columns=["Event", "Gender", "Swim Date", "Swimmer Country"], inplace=True)
df.rename(columns={"Swim Time": "Results", "Meet": "Competition"}, inplace=True)
df.sort_values(by="Year", ascending=False)
df.head()
```

```python
df.to_csv("output/recordes_tratados_homens.csv", index=False)
```
